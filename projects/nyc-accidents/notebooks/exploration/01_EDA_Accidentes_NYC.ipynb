## Fase 2 Investigar nulos y valores
**Objetivo** Ya conociendo las columnas vamos a investigar por columna cuales tienen registros nulos

````python

def resumen_nulos(df):
    """
    Crea un resumen completo de valores nulos en el DataFrame
    
    Args:
        df: DataFrame a analizar
    
    Returns:
        DataFrame con estadísticas de nulos
    """
    total = len(df)
    
    nulos = df.isnull().sum()
    porcentaje = (nulos / total * 100).round(2)
    no_nulos = df.count()
    tipos = df.dtypes
    valores_unicos = df.nunique()
    
    resumen = pd.DataFrame({
        'Columna': df.columns,
        'Tipo': tipos.values,
        'Total Filas': total,
        'No Nulos': no_nulos.values,
        'Nulos': nulos.values,
        '% Nulos': porcentaje.values,
        'Valores Únicos': valores_unicos.values
    })
    
    # Ordenar por % de nulos (mayor a menor)
    resumen = resumen.sort_values('Columna', ascending=True)
    resumen.index = range(1, len(resumen) + 1)
    
    return resumen

# Usamos la función 
print(60*"=")
print("Análisis de nulos del data set por columnas:")
print(resumen_nulos(df))
print(60*"=")
print("Primeras 10 filas:")
print(df.head(10))
````
