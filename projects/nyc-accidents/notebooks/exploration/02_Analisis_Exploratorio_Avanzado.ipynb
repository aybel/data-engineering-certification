```python

df=df.sort_values('CRASH DATE', ascending=True)
```

```python
def resumen_accidente_por_ciudad(df):
    """
        Regresa un df con la ciudad o localidad dentro de nyc y el número de accidentes que ocurrieron 
        Resp:
        {
                {
                    localidad:QUEENS,
                    total_accidentes:150,
                    estatus:alto
                },
            {
                {
                    localidad:BROOKLYN,
                    total_accidentes:5,
                    estatus:BAJO
                }
            }
        }
    """
    # total de accidentes por localidad
    df_resumen=df['BOROUGH'].value_counts().reset_index()
    df_resumen.columns = ['localidad', 'total_accidentes']
    df_resumen['percentil'] = df_resumen['total_accidentes'].rank(pct=True) * 100

    #Establecemos el estatus para el mapa de calor
    condiciones=[
        df_resumen['percentil']<50,
        (df_resumen['percentil']>=50)  & (df_resumen['percentil']<75)
    ]

    valores=['BAJO','ALTO']
    df_resumen['estatus']=np.select(condiciones, valores, default='ALTO')

    return df_resumen

df_resumen=resumen_accidente_por_ciudad(df)
print(60*"=")
print("Resumen de total de accidentes por localidad")
total_accidentes=df_resumen['total_accidentes'].sum()
print(f'Total de accidentes en Nueva york:{total_accidentes}')
print(df_resumen.head())
```
