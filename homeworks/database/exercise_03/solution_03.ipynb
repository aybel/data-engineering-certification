```python
#importar librerias
import sqlite3
connection=sqlite3.connect("../tienda.db")
# Creamos el cursor
cur = connection.cursor()
```

```python
#Consulta los productos conprecio mayor a 300
# 6. CONSULTAR Y MOSTRAR
print("=" * 50)
print("📊 Productos con precio mayor a 300")
print("=" * 50)

cur.execute("""
SELECT * FROM productos
WHERE productos.precio>300
""")
for row in cur.fetchall():
    contador+=1
    print(f"ID: {row[0]} | {row[1]} | ${row[2]:.2f} | Stock: {row[3]}")
print(f"Total de productos: {contador}")
connection.close()
```
