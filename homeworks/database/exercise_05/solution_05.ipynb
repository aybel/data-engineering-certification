```python
def mostrar_productos(cursor):
        """
        Función para mostrar los datos de la tabla productos
        recibe como parametro el cursor
        """
        print("=" * 50)
        print("📊 Lista de productos")
        print("=" * 50)
        cursor.execute("""
              SELECT * 
               FROM
                productos
        """)
        #print(cursor.description)
        for row in cursor.fetchall():
               print(f"ID: {row[0]} | {row[1]} | ${row[2]:.2f} | Stock: {row[3]}")
        
```

```python
def eliminar_producto(cursor, id_producto):
        """
        Función para eliminar un producto de la tabla productos
        recibe como parametro el cursor y el id del producto a eliminar
        """
        cursor.execute("""
                DELETE FROM productos
                WHERE id = ?
        """, (id_producto,))
        print(f"Producto con ID {id_producto} eliminado.")
```

```python
def main():
    #importar librerias
    import sqlite3
    #conexion a la base de datos
    connection=sqlite3.connect("../tienda.db")
    # Creamos el cursor
    cursor = connection.cursor()
    mostrar_productos(cursor)
    eliminar_producto(cursor,1)
    mostrar_productos(cursor)
    connection.close()



if __name__ == "__main__":
    main()
```
