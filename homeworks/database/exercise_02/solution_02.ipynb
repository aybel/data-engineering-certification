```python
#importar librerias
import sqlite3
connection=sqlite3.connect("../tienda.db")
# Creamos el cursor
cur = connection.cursor()
# creamos la tabla 
cur.execute("""
    CREATE TABLE IF NOT EXISTS productos (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nombre TEXT NOT NULL,
        precio REAL,
        stock INTEGER
    )
""")
# insertamos datos
rows = [
    ("Laptop",15000.00,10),
    ("Mouse",250.00,50),
    ("Teclado",450.00,30),
]
# borramos para no duplicar registros
cur.execute("DELETE FROM productos")
# insertamos los 3 
cur.executemany("INSERT INTO productos(nombre,precio,stock) VALUES (?,?,?)",rows)
connection.commit()

# 6. CONSULTAR Y MOSTRAR
print("=" * 50)
print("📊 PRODUCTOS EN LA BASE DE DATOS")
print("=" * 50)

cur.execute("SELECT * FROM productos")
contador = 0  
for row in cur.fetchall():
    contador+=1
    print(f"ID: {row[0]} | {row[1]} | ${row[2]:.2f} | Stock: {row[3]}")
print(f"Total de productos: {contador}")
connection.close()
```
