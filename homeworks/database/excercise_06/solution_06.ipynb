```python
import sqlite3
connection=sqlite3.connect("../tienda.db")
# Creamos el cursor
cur = connection.cursor()

# eliminamos la tabla si existe
cur.execute("DROP TABLE IF EXISTS categorias")

# creamos la tabla 
cur.execute("""
    CREATE TABLE IF NOT EXISTS categorias (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nombre TEXT NOT NULL
    )
""")
# insertamos datos
rows = [
    ("Electrónicos",),
    ("Accesorios",),
    ("Periféricos",),
]

# Actualizamos la tabla con los datos
cur.executemany("INSERT INTO categorias (nombre) VALUES (?)",rows)
connection.commit()
# Creamos el campo categeoria_id
cur.execute("ALTER TABLE productos ADD COLUMN categoria_id INTEGER REFERENCES categorias(id)")
# Actualizamos la tabla 
connection.commit()
# Actualizamos los productos 

cur.execute("""
    UPDATE productos 
    SET categoria_id = ? 
    WHERE id = ?
    """, (1, 1))
cur.execute("""
    UPDATE productos 
    SET categoria_id = ? 
    WHERE id = ?
    """, (2, 2))
cur.execute("""
    UPDATE productos 
    SET categoria_id = ? 
    WHERE id = ?
    """, (3, 3))
connection.commit()

connection.close()
print("\n🔒 Conexión cerrada")
```

```python
print("\n📋 PRODUCTOS CON CATEGORÍAS:")
cur.execute("""
    SELECT 
        p.id,
        p.nombre,
        p.precio,
        p.stock,
        c.nombre as categoria
    FROM productos p
    LEFT JOIN categorias c ON p.categoria_id = c.id
    ORDER BY p.id
""")

for row in cur.fetchall():
    categoria = row[4] if row[4] else "Sin categoría"
    print(f"  ID: {row[0]} | {row[1]} | ${row[2]:.2f} | Stock: {row[3]} | Categoría: {categoria}")
```
